```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
df=pd.read_csv(r"C:\Users\sandr\Downloads\data_season.csv")
print(df)
```

```python
# Display the average price for each crop.
avg_price = df.groupby("Crops")["price"].mean()
plt.bar(avg_price.index, avg_price.values)
plt.title("Average Price by Crop")
plt.xlabel("Crops")
plt.ylabel("Average Price")
plt.xticks(rotation=45)
plt.show()
```

```python
# Display the distribution of rainfall.
plt.hist(df["Rainfall"], bins=5)
plt.title("Rainfall Distribution")
plt.xlabel("Rainfall")
plt.ylabel("Frequency")
plt.show()
```

```python
# Display the relationship between rainfall and temperature.
plt.scatter(df["Rainfall"], df["Temperature"])
plt.title("Rainfall vs Temperature")
plt.xlabel("Rainfall")
plt.ylabel("Temperature")
plt.show()
```

```python
# Display the average price for each crop.
avg_price = df.groupby("Crops")["price"].mean()
plt.bar(avg_price.index, avg_price.values)
plt.title("Average Price by Crop")
plt.xlabel("Crops")
plt.ylabel("Average Price")
plt.xticks(rotation=45)
plt.show()
```

```python
# Display the number of records for each season.
sns.countplot(data=df, x="Season")
plt.title("Number of Records by Season")
plt.show()
```

```python
# Compare rainfall across seasons.
sns.boxplot(data=df, x="Season", y="Rainfall")
plt.title("Rainfall by Season")
plt.show()
```

```python
# Display the relationship between rainfall and price.
sns.scatterplot(data=df, x="Rainfall", y="price")
plt.title("Rainfall vs Price")
plt.show()
```

```python
# Display the correlation between numerical columns.
sns.heatmap(df.corr(numeric_only=True))
plt.title("Correlation Heatmap")
plt.show()
```

```python
# Display the distribution of crops using a pie chart.
crop_count = df["Crops"].value_counts()
plt.pie(crop_count.values, labels=crop_count.index, autopct="%1.1f%%")
plt.title("Crop Distribution")
plt.show()
```

```python
# Display the number of records where rainfall is greater than 2500.
data = df[df["Rainfall"] > 2500]["Season"].value_counts()
plt.bar(data.index, data.values)
plt.title("Seasons with Rainfall Greater Than 2500")
plt.xlabel("Season")
plt.ylabel("Count")
plt.show()
```

```python
# Display the crop count where rainfall is greater than 2000 and price is greater than 50000.
data = df[(df["Rainfall"] > 2000) & (df["price"] > 50000)]["Crops"].value_counts()
plt.bar(data.index, data.values)
plt.title("Crops with High Rainfall and High Price")
plt.xlabel("Crops")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
# Display the average price of crops where temperature is greater than 30.
data = df[df["Temperature"] > 30].groupby("Crops")["price"].mean()
plt.bar(data.index, data.values)
plt.title("Average Price of Crops at Temperature Above 30")
plt.xlabel("Crops")
plt.ylabel("Average Price")
plt.xticks(rotation=45)
plt.show()
```
